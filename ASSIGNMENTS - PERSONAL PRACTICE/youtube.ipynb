## FOR LOOP

Print the 7 times table from 1 to 12 using a for loop

```python
numbers = range(1,13)
for number in numbers:
    multiples_of_7 = (7 * number)
    print(f"7 x {number} = {multiples_of_7}")
```

Print a left aligned pyramid of stars with 6 rows using a for loop

```python
stars = range(1,7)

for star in stars:
    rows = ("*" * star)
    print (rows)
```

- Check whether any file names appears more than once
- Print duplicate found if a duplicate exists otherwise print all files are unique

```python
# Check whether any file names appears more than once
# Print duplicate found if a duplicate exists otherwise print all files are unique

file_list = [
    "report.csv",
    "data.xlsx",
    "summary.docx",
    "report.csv",
    "data.csv"
]

# use a for loop to iterate the list
for file in file_list:
    file_count = file_list.count(file)
    file_count = 0
    if file_count > 1:
        print(f"Duplicate found: {file}")
        break
    else:
        print(f"All files are unique")
```

## WHILE LOOP


Extend the loop below with the requirements below
- Limit user's response to 3 attempts in total. If yes is not given as a response, do not ask anymore
- If the user types yes within the 3 attempts, print "Glad we are on the same page"
- Otherwise, print "3 strikes, You are Out!"

```python
while True:
    answer = input("Do you agree? Yes/No: ")
    if answer == "yes":
        break

print("Thank You")
```

```python
# MY ANSWER
# # Limit user's response to 3 attempts

answer_attempt = 0
# answer_count = 0
while True:
    answer = input("Do you agree? Yes/No: ")
    # If answer attempt is > 3 and yes is not given as a response, print nothing
    if answer_attempt > 3 and answer != "yes":
        answer_attempt += answer
        break
    print()
    # If answer attempt <= 3 and answer = yes, print "Glad we are on the same page"
    if answer_attempt <= 3 and answer == "yes":
        print("Glad we are on the same page")
        break
    # If answer attempt >= 3 and answer != yes, print ""3 strikes! You are out!""
    else:
        print("3 strikes! \n You are out!")
    
print("Thank You")
```

```python
# CORRECTION
# Limit user's response to 3 attempts

answer_attempt = 0

while answer_attempt < 3:
    answer = input("Do you agree? Yes/No: ")
    if answer == "yes":
        print("Glad we are on the same page")
        break
    answer_attempt += 1
else:
    print("3 strikes! \n You are out!")
    
print("Thank You")
```
