```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score 

#Load historical data
customer = pd.read_csv("customer_churn.csv")

# To view the no. of rows and columns
print("Dataset shape:", customer.shape)
print()

# To print the first 8 rows of the file
print(customer.head(8))
```

```python
# Separate features (X) from the label(y)
# Features = What we know about the applicant
# Label = The outcome we want to predict

X = customer[["monthly_spend", "num_complaints"]]
y = customer["churned"]

# Note Churned customers means that they are leaving and non-returning customers
# Unchurned means that they are returning customer

print("Features (X) - 3 columns, 20 rows: ")

# Print first 3 rows
print(X.head(3))
# print(X.head)
print()
# Meaning of the Values
# (True which is 1) (Customer = Not returning): The customer has churned. They cancelled their subscription, closed their account, or stopped buying from the business.
# (False which is 0) ( Customer = Returning): The customer is retained. They are active, loyal, or still using the service.

print("Label (y) - what we want to predict: ")
print(y.values)
print(f"Churned customer: {y.sum()} | Unchurned customer: {(y ==0).sum()}")
```

```python
# Split: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 0.2, random_state = 10
)

print(f"Training set: {len(X_train)}")
```
