## FOR LOOP

Print the 7 times table from 1 to 12 using a for loop

```python
numbers = range(1,13)
for number in numbers:
    multiples_of_7 = (7 * number)
    print(f"7 x {number} = {multiples_of_7}")
```

Print a left aligned pyramid of stars with 6 rows using a for loop

```python
stars = range(1,7)

for star in stars:
    rows = ("*" * star)
    print (rows)
```

```python
# Check whether any file names appears more than once
# Print duplicate found if a duplicate exists otherwise print all files are unique

file_list = [
    "report.csv",
    "data.xlsx",
    "summary.docx",
    "report.csv",
    "data.csv"
]

# use a for loop to iterate the list
for file in file_list:
    file_count = file_list.count(file)
    #file_count = 0
    if file_count > 1:
        print(f"Duplicate found: {file}")
        break
    else:
        print(f"All files are unique")
```
