# Practice Question — Topic 4

- Ask for: exam score (float), coursework score (float), attendance percentage (float, 0–100)
- Calculate the final score: exam is 60%, coursework is 40%
- Add a 5% bonus if attendance >= 90%
- Determine pass/fail: final score >= 50 AND attendance >= 60%

- Display: final score (rounded to 2 decimals), pass/fail status, and the attendance bonus applied (True/False)

# Class Activity 5.3a
Task: Build a BMI classifier. Ask for weight (kg) and height (m). Calculate BMI = weight / height². Classify: Underweight (<18.5), Normal (18.5–24.9), Overweight (25–29.9), Obese (>=30).

```python
# To build a BMI (Body Mass Index) classifier
weight = float(input("What is your weight in kg?"))
height = float(input("How tall are you in metres"))

print("Weight (in kg):", weight)
print("Height (in m):", height)

# Body mass index formula
bmi = ((weight) / (height **2))
print("Your BMI is:", bmi)

if bmi < 18.5:
    print("You are underweight!")
elif 18.5 < bmi < 25:
    print("Your body mass index is perfect!")
elif 25 < bmi < 30:
    print("You are overweight!") 
else:
    print("You are obese!")
```
