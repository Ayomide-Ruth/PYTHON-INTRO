# Practice Question — Topic 4
Task: Build a grade calculator that uses all operator types.

- Ask for: exam score (float), coursework score (float), attendance percentage (float, 0–100)
- Calculate the final score: exam is 60%, coursework is 40%
- Add a 5% bonus if attendance >= 90%
- Determine pass/fail: final score >= 50 AND attendance >= 60%
- Use += when applying the bonus
- Display: final score (rounded to 2 decimals), pass/fail status, and the attendance bonus applied (True/False)

```python
# Create a border
print("=" * 40)
print("GRADE CALCULATOR")
print("=" * 40)

# Assign variable names to values
exam_score = float(input("What is your exam score?"))
coursework_score = float(input("What was your course work score?"))
attendance = float(input("What is your percentage attendance? 0 - 100%"))

# 60% of exam and 40% of test
sum_score = (exam_score * 0.6) + (coursework_score * 0.4)

# Check for attendance bonus
attendance_bonus = attendance >=90

# if attendance is >= 90, add 0.05 using +=
if attendance_bonus:
    sum_score += sum_score * 0.05

# Calculate final score
final_score = (sum_score + attendance_bonus)

# Use the round () function to round final score to 2 decimal place
round(final_score, 2)

print("Final Score:", final_score)

# Check Status
pass_ = "Pass"
fail_ = "Fail"

if final_score >= 50 and attendance >=60:
    print("Status:", pass_)
else:
    print("Status:", fail_)

# pass_ = final_score >= 50 and attendance >=60 or bonus_applied
# fail_ = final_score <=49 and attendance <=59 or bonus_applied

if attendance >= 90:
    print("Bonus Applied: True")
else:
    print("Bonus Applied: False")
```

# Class Activity 5.3a
Task: Build a BMI classifier. Ask for weight (kg) and height (m). Calculate BMI = weight / height². Classify: Underweight (<18.5), Normal (18.5–24.9), Overweight (25–29.9), Obese (>=30).

```python
# To build a BMI (Body Mass Index) classifier
weight = float(input("What is your weight in kg?"))
height = float(input("How tall are you in metres"))

print("Weight (in kg):", weight)
print("Height (in m):", height)

# Body mass index formula
bmi = ((weight) / (height **2))
print("Your BMI is:", bmi)

if bmi < 18.5:
    print("You are underweight!")
elif 18.5 < bmi < 25:
    print("Your body mass index is perfect!")
elif 25 < bmi < 30:
    print("You are overweight!") 
else:
    print("You are obese!")
```

- Create a list of 5 products where each product is a tuple: (name, price, quantity)
- Add 2 more products using .append()
- Remove one product by unpacking its name and using .remove()
- Iterate using enumerate() to display a numbered inventory
- Calculate: total inventory value (price × quantity for all), most expensive item

```python
# List all products in an inventory
inventory = [ 
    ("rice", 4500, 35),
    ("beans", 440, 25),
    ("cassava", 1000, 16),
    ("yam", 2500, 75)
    ]

# Add 2 more products using .append()
inventory.append(
    ("banana", 3500, 19)   
)

inventory.append(
    ("apple", 500, 100)   
)

print(inventory)

# Remove selected product by unpacking
for product in inventory:
    name, price, qty = product
    if product[0] == "cassava":
        inventory.remove(product)

print(inventory)

# Print enumerated inventory using .enumerate()
for num, product in enumerate(inventory, start = 1):
    print(f"{num} - {product}")

# Get total inventory amount
total_price = 0

for product in inventory:
    name, price, qty = product
    total_price += product[1] * product[2]

print(f"total inventory price = {total_price}")     #Note in using the f function, variable names are put in curly brackets

# Get most expensive product
most_expensive = inventory[2]

for product in inventory:
    if product [1] > most_expensive[1]:
        most_expensive = product
print(most_expensive)
print(most_expensive[0:-1])
```

Build a Course Enrollment Tracker:

- Create a dictionary mapping course names to a set of enrolled students
- Add at least 2 more students to one course using .add() on the set
- Use set operations to find: students in both Course A and Course B (intersection), students in Course A only (difference)
- Iterate the dictionary with .items() to print each course and student count
- Find which course has the most students

```python
# enrolment_list = {
#     "course": "English", "Names": eng_stu
#     eng_stu = {"Funmi", "Shade", "Victor"}}

# math_stu = {"Funmi", "Shade", "Victor"},
# eng_stu = {"Hope", "Bolu", "Grace"},
# bio_stu ={"Favour", "Moji", "Tofunmi"}
```

```python
# 
enrolment_list = {
"Maths": {"Funmi", "Shade", "Victor"},
"English": {"Hope", "Victor", "Grace"},
#"Biology": {"Favour", "Moji", "Tofunmi"}
}

# Add at least 2 students to one course using .add()
enrolment_list["Maths"].add("Peter")
enrolment_list["Maths"].add("Alice")
print(enrolment_list)

# Finding students in both courses using .intersection()
student_in_both = enrolment_list["Maths"].intersection(enrolment_list["English"])
print(f"List of students in maths and English class {student_in_both}")

# Finding students in course A using .difference()
student_in_maths = enrolment_list["Maths"].difference(enrolment_list["English"])
print(f"List of students in maths class {student_in_maths}")

# To iterate the dictionary with .items() to print each course and student count
for course, students in enrolment_list.items():
    print(f"{course}: {students}")

# Find which course has the most students
no_of_maths_stu = len(enrolment_list["Maths"])
no_of_eng_stu = len(enrolment_list["English"])

if no_of_maths_stu > no_of_eng_stu:
    print(f"Maths has the most students")
else:
    print("English has the most students")

# find the course with the highest student count
highest_count = 0
winning_course = ""

for course, students in enrolment_list.items():
    if len(students) > highest_count:
        winning_course = course
print(winning_course)
```
