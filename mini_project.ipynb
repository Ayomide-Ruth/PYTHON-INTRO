- Every section of code must be correctly indented

```python
# Assign variable names to collect users information
name = input("Input your First and Last name:")
age = int(input("What year were you born"))
city = input("What city do you live in?")
time = int(input("What time of the day are you currently in? 24-hour format"))

# To creat borders, we use the code below:
print("=" * 40)

# To get what to display what time of the day it is, we use the conditional statements below:
if 0 <= time  <= 11:
    print ("Good morning!")
elif 12 <= time <= 15:
    print("Good afternoon!")
else:
    print("Good evening!")

# To create a title for the greeting system

print("=" * 40)

# To calculate the user's age based on the year the user was born, we use;
cal_age = int((2026 - age))

# To display our data, we use the print() function
print("My name is", name + ".")
print("I am", cal_age, "years old" + ".")
print("I reside in the city of", city + ".")

# To close the border
print("=" * 40)
```
